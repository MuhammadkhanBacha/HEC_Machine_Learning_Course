```python
from sklearn.linear_model import LinearRegression
import numpy as np

X = np.array([[1], [2], [3], [4], [5]])  # must be 2D
y = np.array([2, 4, 5, 4, 5])

model = LinearRegression() # empty object initialized
model.fit(X, y) # formula-> it solves the xtx 

model.predict([[6]])
```

```python
model.coef_
```

```python
model.intercept_
```

```python
m = model.coef_
b = model.intercept_
m * 6 + b
```
