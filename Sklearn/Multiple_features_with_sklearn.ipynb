```python
from sklearn.linear_model import LinearRegression
import numpy as np
```

```python
# Example: predicting y from 2 features
X = np.array([[1, 2], [2, 1], [3, 4], [4, 3], [5, 5]])
y = np.array([5, 4, 10, 9, 13])
```

```python
model = LinearRegression()
model.fit(X,y)
model.coef_
model.intercept_
```

```python
model.predict([[6, 7]])
```
